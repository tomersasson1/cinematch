# Exploratory Data Analysis – MovieLens

This notebook explores the MovieLens ratings and movies data: distributions, popular genres, and sparsity of the user–item matrix.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

import pandas as pd
from recommender.data_loading import load_movielens
from recommender import config

ratings, movies = load_movielens()
print("Ratings shape:", ratings.shape)
print("Movies shape:", movies.shape)
print(ratings.head())
print(movies.head())
```

```python
# Ratings per user and per movie
user_counts = ratings.groupby(config.USER_ID_COL)[config.ITEM_ID_COL].count()
item_counts = ratings.groupby(config.ITEM_ID_COL)[config.USER_ID_COL].count()
user_counts.describe(), item_counts.describe()
```

```python
# Sparsity
n_users = ratings[config.USER_ID_COL].nunique()
n_items = ratings[config.ITEM_ID_COL].nunique()
observed = len(ratings)
total = n_users * n_items
print(f"Users: {n_users}, Items: {n_items}, Observed: {observed}, Sparsity: {observed/total:.6f}")
```

```python
# Genres
def split_genres(s):
    if pd.isna(s) or s == "(no genres listed)": return []
    return [g.strip() for g in str(s).split("|") if g]
movies["genre_list"] = movies[config.GENRES_COL].apply(split_genres)
all_genres = []
for gs in movies["genre_list"]: all_genres.extend(gs)
pd.Series(all_genres).value_counts()
```
